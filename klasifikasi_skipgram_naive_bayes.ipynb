## 4. Konsep Skip-gram (demo 3 kalimat)

**Word2Vec** mengubah setiap kata menjadi vektor angka berdimensi tetap. Kata yang sering muncul di **konteks yang mirip** akan punya vektor yang berdekatan.

**Skip-gram**: satu **kata target** dipakai untuk menebak **kata-kata di sekitarnya (konteks)**.

| Parameter | Arti |
|---|---|
| `vector_size` | panjang vektor tiap kata (jumlah dimensi) |
| `window` | berapa kata di kiri dan kanan target yang dianggap konteks |
| `epochs` | berapa kali model membaca seluruh data latih |
| `sg=1` | pakai Skip-gram (`sg=0` = CBOW) |
| `min_count` | kata yang muncul kurang dari ini diabaikan |

Contoh `window=2` pada kalimat `harga saham bank naik hari ini`, target `saham`:

$$\text{saham} \rightarrow \{\text{harga},\ \text{bank},\ \text{naik}\}$$

**Skip-gram vs CBOW** (untuk jawaban jika ditanya kenapa memilih Skip-gram):

| | Skip-gram | CBOW |
|---|---|---|
| Arah | 1 target → banyak konteks | banyak konteks → 1 target |
| Kata jarang | lebih baik | kurang baik |
| Data kecil | cocok | kurang cocok |
| Kecepatan | lebih lambat | lebih cepat |

````python
# Dataset mini (sudah huruf kecil & tanpa tanda baca)
kalimat_demo = [
    "harga saham bank naik hari ini",
    "investor membeli saham bank besar",
    "pemain bola mencetak gol indah",
]
korpus_demo = [k.split() for k in kalimat_demo]
VECTOR_SIZE_DEMO, WINDOW_DEMO = 5, 2

def pasangan_target_konteks(tokens, window):
    # Tiap kata jadi target, lalu dipasangkan dengan kata di kiri/kanan sejauh `window`
    pasangan = []
    for i, target in enumerate(tokens):
        for j in range(max(0, i - window), min(len(tokens), i + window + 1)):
            if j != i:
                pasangan.append((target, tokens[j]))
    return pasangan

print("Kalimat 1 :", korpus_demo[0], "\n")
pasangan = pasangan_target_konteks(korpus_demo[0], WINDOW_DEMO)
for target in korpus_demo[0]:
    print(f"{target:8s} -> {[k for t, k in pasangan if t == target]}")

total = sum(len(pasangan_target_konteks(k, WINDOW_DEMO)) for k in korpus_demo)
print(f"\nTotal pasangan (target, konteks) pada 3 kalimat : {total}")
pd.DataFrame(pasangan, columns=["target", "konteks"]).head(10)
````

````python
# Latih Skip-gram kecil, lalu lihat vektor kata "saham"
model_demo = Word2Vec(sentences=korpus_demo, vector_size=VECTOR_SIZE_DEMO, window=WINDOW_DEMO,
                      min_count=1, sg=1, epochs=100, seed=SEED, workers=1)

print("Vektor kata 'saham' :", np.round(model_demo.wv["saham"], 4))
print("Bentuk vektor       :", model_demo.wv["saham"].shape)

def cosine_similarity(a, b):
    return float(np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b)))

print("\nKemiripan (cosine) dengan 'saham':")
for kata in ["bank", "investor", "gol", "pemain"]:
    print(f"  saham vs {kata:9s}: {cosine_similarity(model_demo.wv['saham'], model_demo.wv[kata]):.4f}")
````

**Cara membaca output:**
- Tiap kata menjadi vektor sepanjang `vector_size` (di sini 5 angka).
- **Cosine similarity** mengukur kemiripan arah dua vektor: dekat `1` = mirip, dekat `0` = tidak berhubungan, negatif = berlawanan.

$$\cos(\mathbf{a},\mathbf{b}) = \frac{\mathbf{a}\cdot\mathbf{b}}{\lVert\mathbf{a}\rVert\,\lVert\mathbf{b}\rVert}$$

## 5. Preprocessing P1–P4

Semua skema **selalu** diawali dua langkah dasar:
1. **Case folding** — ubah ke huruf kecil (`Saham` → `saham`)
2. **Hapus tanda baca** — (`Jakarta, harga` → `jakarta harga`)

lalu dipecah menjadi token (tokenisasi). Perbedaan antar skema hanya pada langkah tambahan:

| Skema | Langkah | Peran |
|---|---|---|
| **P1** `P1_lower_tandabaca` | huruf kecil + hapus tanda baca | **dipakai untuk melatih model final** |
| P2 `P2_stopword` | P1 + hapus stopword | eksperimen |
| P3 `P3_stemming` | P1 + stemming | eksperimen |
| P4 `P4_stopword_stemming` | P1 + stopword + stemming | eksperimen |

- **Stopword** = kata umum yang kurang bermakna (`yang`, `dan`, `untuk`) — daftar dari Sastrawi.
- **Stemming** = mengubah kata ke bentuk dasar (`menyebut` → `sebut`) — algoritma Sastrawi (Nazief–Adriani).

Angka dan token 1 huruf **tidak dibuang** pada semua skema, supaya perbedaannya hanya pada langkah di tabel.

````python
_cache_stem = {}   # stemming lambat, jadi hasilnya disimpan per kata unik

def preprocess(teks, hapus_stopword=False, stemming=False):
    # Teks mentah -> list token. Langkah dasar (P1) selalu jalan.
    teks = str(teks).lower()                 # 1. huruf kecil
    teks = re.sub(r"[^\w\s]", " ", teks)     # 2. hapus tanda baca
    tokens = teks.split()                    # 3. tokenisasi
    if hapus_stopword:                       # tambahan (P2, P4)
        tokens = [t for t in tokens if t not in STOPWORDS_ID]
    if stemming:                             # tambahan (P3, P4)
        for i, t in enumerate(tokens):
            if t not in _cache_stem:
                _cache_stem[t] = STEMMER.stem(t)
            tokens[i] = _cache_stem[t]
    return tokens

PREP_CONFIGS = {
    "P1_lower_tandabaca":   dict(hapus_stopword=False, stemming=False),
    "P2_stopword":          dict(hapus_stopword=True,  stemming=False),
    "P3_stemming":          dict(hapus_stopword=False, stemming=True),
    "P4_stopword_stemming": dict(hapus_stopword=True,  stemming=True),
}
MODEL_PREP = "P1_lower_tandabaca"     # skema yang dipakai model final

prep_results = {}
for nama, cfg in PREP_CONFIGS.items():
    print("Memproses", nama, "...")
    prep_results[nama] = df_berita["teks"].apply(preprocess, **cfg)

print("\nTEKS ASLI :", df_berita.loc[0, "teks"][:200])
for nama in PREP_CONFIGS:
    print(f"\n[{nama}] ->", prep_results[nama][0][:12])

print("\nUkuran kosakata unik tiap skema:")
for nama, serial in prep_results.items():
    print(f"  {nama:22s}: {len(set(t for dok in serial for t in dok))} kata")
````

## 6. Split Data Latih / Uji

Data dibagi **80% latih dan 20% uji** secara *stratified* (proporsi `sport` : `finance` tetap sama). Label diubah ke angka: `sport` = 0, `finance` = 1.

Yang disimpan adalah **indeks** (`idx_train`, `idx_test`). Dengan begitu semua skema preprocessing diuji pada **dokumen uji yang persis sama**, sehingga perbandingannya adil.

**Aturan penting:** data uji **tidak boleh** dipakai untuk melatih Word2Vec maupun Naive Bayes (kalau dipakai = *data leakage*).

````python
PETA_LABEL = {"sport": 0, "finance": 1}
if not df_berita["label"].isin(PETA_LABEL).all():
    raise ValueError(f"Label tak dikenal: {df_berita['label'].unique().tolist()}")

y = df_berita["label"].map(PETA_LABEL).values

idx_train, idx_test = train_test_split(np.arange(len(df_berita)), test_size=0.2,
                                       random_state=SEED, stratify=y)
y_train, y_test = y[idx_train], y[idx_test]

print("Data latih :", len(idx_train), "| Data uji :", len(idx_test))
print("Kelas data latih :", np.bincount(y_train), "| Kelas data uji :", np.bincount(y_test))
````

## 7. Eksperimen

### 7.1 Fungsi inti: Skip-gram → vektor dokumen → Gaussian Naive Bayes

Satu fungsi `evaluasi()` dipakai **di semua tempat** (eksperimen, grid search, model final), jadi cukup dijelaskan sekali:

1. **Latih Skip-gram hanya dari dokumen latih.**
2. **Vektor dokumen = rata-rata vektor semua kata di dokumen itu:**

$$\mathbf{d}_i = \frac{1}{n_i}\sum_{k=1}^{n_i}\mathbf{v}_{t_{ik}}$$

   Kata yang tidak ada di kosakata dilewati; dokumen tanpa kata dikenal diberi vektor nol.
3. **Gaussian Naive Bayes** dilatih pada vektor dokumen. Dipilih Gaussian karena fiturnya bilangan real (bisa negatif).

Fungsi `cv_skipgram()` mengulang langkah di atas dengan **cross-validation** pada data latih saja: data latih dibagi 5 bagian, bergantian 4 bagian untuk melatih dan 1 bagian untuk menguji. Hasilnya rata-rata 5 akurasi — lebih stabil daripada satu kali split dan **tidak menyentuh data uji**.

````python
def dokumen_ke_vektor(tokens, w2v):
    # Rata-rata vektor semua kata dalam dokumen (kata di luar kosakata dilewati)
    vektor = [w2v.wv[t] for t in tokens if t in w2v.wv]
    return np.mean(vektor, axis=0) if vektor else np.zeros(w2v.vector_size)

def evaluasi(tok_latih, y_latih, tok_uji, y_uji, vector_size=100, window=5, epochs=20, var_smoothing=1e-9):
    # Latih Skip-gram + GaussianNB pada data latih, uji pada data uji.
    w2v = Word2Vec(sentences=tok_latih, vector_size=vector_size, window=window,
                   min_count=1, sg=1, epochs=epochs, seed=SEED, workers=1)
    X_latih = np.array([dokumen_ke_vektor(t, w2v) for t in tok_latih])
    X_uji   = np.array([dokumen_ke_vektor(t, w2v) for t in tok_uji])
    nb = GaussianNB(var_smoothing=var_smoothing).fit(X_latih, y_latih)
    y_pred = nb.predict(X_uji)
    return accuracy_score(y_uji, y_pred), y_pred, w2v, nb

def cv_skipgram(tok_latih, y_latih, k=5, **param):
    # Rata-rata akurasi k-fold cross-validation (hanya memakai data latih)
    skf = StratifiedKFold(n_splits=k, shuffle=True, random_state=SEED)
    skor = []
    for tr, va in skf.split(np.zeros(len(y_latih)), y_latih):
        acc, _, _, _ = evaluasi([tok_latih[i] for i in tr], y_latih[tr],
                                [tok_latih[i] for i in va], y_latih[va], **param)
        skor.append(acc)
    return float(np.mean(skor))

def ambil_token(nama_skema):
    # Token skema tertentu -> (token latih, token uji)
    tokens = prep_results[nama_skema].tolist()
    return [tokens[i] for i in idx_train], [tokens[i] for i in idx_test]
````

### 7.2 Pengaruh preprocessing (P1–P4)

Keempat skema diuji dengan **parameter Skip-gram yang sama** (`vector_size=100`, `window=5`, `epochs=20`) agar adil. Dua angka dilaporkan:
- **akurasi_cv** — rata-rata cross-validation pada data latih
- **akurasi_uji** — akurasi pada data uji

Tujuannya hanya melihat apakah stopword / stemming membantu. **Model final tetap memakai P1**.

````python
PARAM_TETAP = dict(vector_size=100, window=5, epochs=20)

hasil_prep = []
for nama in PREP_CONFIGS:
    tok_tr, tok_te = ambil_token(nama)
    acc_cv = cv_skipgram(tok_tr, y_train, **PARAM_TETAP)
    acc_uji = evaluasi(tok_tr, y_train, tok_te, y_test, **PARAM_TETAP)[0]
    hasil_prep.append({"skema": nama, "akurasi_cv": acc_cv, "akurasi_uji": acc_uji})
    print(f"{nama:22s} | CV = {acc_cv:.4f} | uji = {acc_uji:.4f}")

df_prep = pd.DataFrame(hasil_prep)

ax = df_prep.set_index("skema")[["akurasi_cv", "akurasi_uji"]].plot(kind="bar", figsize=(8, 4), rot=15)
ax.set_title("Pengaruh Preprocessing (Skip-gram + Gaussian NB, parameter sama)")
ax.set_ylabel("Akurasi")
ax.set_ylim(0, 1.05)
plt.tight_layout()
plt.show()
df_prep
````

### 7.3 Pembanding: One-Hot Encoding + Bernoulli Naive Bayes (pada P1)

**One-Hot Encoding** (Bag-of-Words biner): tiap kata di kosakata menjadi satu kolom; nilainya `1` jika kata muncul di dokumen, `0` jika tidak. Karena fiturnya biner, dipakai **BernoulliNB**.

Berbeda dengan Skip-gram, One-Hot **tidak menangkap kemiripan makna** antar kata (`saham` dan `obligasi` dianggap dua kolom yang tidak berhubungan). Kosakata dibentuk **hanya dari data latih**.

````python
tok_tr_p1, tok_te_p1 = ambil_token(MODEL_PREP)

onehot = CountVectorizer(tokenizer=lambda x: x, preprocessor=lambda x: x,
                         lowercase=False, token_pattern=None, binary=True)
X_tr_oh = onehot.fit_transform(tok_tr_p1)     # kosakata dari data latih saja
X_te_oh = onehot.transform(tok_te_p1)

nb_oh = BernoulliNB().fit(X_tr_oh, y_train)
acc_onehot = accuracy_score(y_test, nb_oh.predict(X_te_oh))

print("Ukuran kosakata One-Hot (jumlah fitur) :", X_tr_oh.shape[1])
print(f"Akurasi uji One-Hot + BernoulliNB (P1)  : {acc_onehot:.4f}")
````

## 8. Pencarian Parameter Otomatis (Optuna, pada P1)

Tujuan: **sistem sendiri** mencari kombinasi parameter yang paling bagus untuk model kita (P1). Kita tidak menulis daftar nilai yang dicoba — kita hanya menentukan **rentang** yang boleh dijelajahi.

| Parameter | Rentang yang dijelajahi |
|---|---|
| `vector_size` | 20 – 200 (kelipatan 10) |
| `window` | 1 – 10 |
| `epochs` | 10 – 100 (kelipatan 10) |
| `var_smoothing` (Gaussian NB) | 1e-9 – 1e-1 (skala log) |

**Cara kerja (sederhana):**
1. Optuna memilih satu kombinasi parameter dari rentang di atas.
2. Kombinasi itu dinilai dengan **cross-validation 5-fold pada data latih** (`cv_skipgram`).
3. Optuna **belajar dari hasil percobaan sebelumnya** (algoritma TPE, sejenis Bayesian optimization), sehingga percobaan berikutnya lebih sering mengarah ke area parameter yang hasilnya bagus. Ini lebih efisien daripada mencoba semua kombinasi satu per satu (Grid Search).
4. Diulang sebanyak `N_TRIAL` kali. Kombinasi dengan **akurasi CV tertinggi** dipilih.

Seperti sebelumnya, pemilihan **tidak memakai data uji**, sehingga data uji tetap murni untuk penilaian akhir.

> Ingin pencarian lebih teliti? Naikkan `N_TRIAL` (makin besar, makin lama). `seed` dibuat tetap supaya hasil bisa diulang.

````python
N_TRIAL = 30   # jumlah percobaan kombinasi parameter yang dilakukan sistem

optuna.logging.set_verbosity(optuna.logging.WARNING)

def objective(trial):
    # Optuna mengusulkan parameter dari rentang, lalu kita kembalikan skor CV-nya
    param = dict(
        vector_size   = trial.suggest_int("vector_size", 20, 200, step=10),
        window        = trial.suggest_int("window", 1, 10),
        epochs        = trial.suggest_int("epochs", 10, 100, step=10),
        var_smoothing = trial.suggest_float("var_smoothing", 1e-9, 1e-1, log=True),
    )
    return cv_skipgram(tok_tr_p1, y_train, **param)

def cetak_progres(study, trial):
    print(f"Percobaan {trial.number + 1:2d}/{N_TRIAL} | CV = {trial.value:.4f} | "
          f"terbaik sejauh ini = {study.best_value:.4f} | {trial.params}")

study = optuna.create_study(direction="maximize", sampler=optuna.samplers.TPESampler(seed=SEED))
study.optimize(objective, n_trials=N_TRIAL, callbacks=[cetak_progres])

BEST_PARAMS = dict(study.best_params)
print("\nParameter terbaik :", BEST_PARAMS)
print(f"Akurasi CV terbaik : {study.best_value:.4f}")
````

````python
# 10 percobaan terbaik + grafik perkembangan skor
df_trial = (study.trials_dataframe()[["number", "value", "params_vector_size", "params_window",
                                      "params_epochs", "params_var_smoothing"]]
              .rename(columns={"number": "percobaan", "value": "akurasi_cv",
                               "params_vector_size": "vector_size", "params_window": "window",
                               "params_epochs": "epochs", "params_var_smoothing": "var_smoothing"})
              .sort_values("akurasi_cv", ascending=False)
              .reset_index(drop=True))

riwayat = study.trials_dataframe().sort_values("number")
plt.figure(figsize=(8, 3.5))
plt.plot(riwayat["number"] + 1, riwayat["value"], "o", alpha=0.5, label="skor tiap percobaan")
plt.plot(riwayat["number"] + 1, riwayat["value"].cummax(), "-", color="crimson", label="terbaik sejauh ini")
plt.xlabel("Percobaan ke-"); plt.ylabel("Akurasi CV")
plt.title("Perkembangan Pencarian Parameter Otomatis")
plt.legend()
plt.tight_layout()
plt.show()
df_trial.head(10)
````

## 9. Model Final: P1 + Parameter Terbaik

Model final dilatih pada **seluruh data latih** (P1, parameter terbaik hasil pencarian otomatis), lalu diuji **satu kali** pada data uji.

**Gaussian Naive Bayes** memakai teorema Bayes dengan asumsi tiap dimensi vektor saling bebas dan berdistribusi normal:

$$\hat{c} = \arg\max_{c}\left[\log P(c) + \sum_{j=1}^{d}\log P(x_j \mid c)\right]$$

**Metrik** (kelas positif = `finance`):

$$\text{Accuracy}=\frac{TP+TN}{TP+TN+FP+FN},\quad \text{Precision}=\frac{TP}{TP+FP},\quad \text{Recall}=\frac{TP}{TP+FN},\quad F_1=2\cdot\frac{P\cdot R}{P+R}$$

````python
akurasi_final, y_pred, w2v_final, nb_final = evaluasi(tok_tr_p1, y_train, tok_te_p1, y_test, **BEST_PARAMS)

print("Skema preprocessing :", MODEL_PREP)
print("Parameter Skip-gram :", BEST_PARAMS)
print(f"Akurasi uji         : {akurasi_final:.4f}")
print("Ukuran kosakata     :", len(w2v_final.wv))

print("\nClassification report:")
print(classification_report(y_test, y_pred, target_names=["sport", "finance"], digits=4))

ConfusionMatrixDisplay.from_predictions(y_test, y_pred, display_labels=["sport", "finance"], cmap="Blues")
plt.title("Confusion Matrix — Skip-gram + Gaussian NB (P1)")
plt.show()
````

## 10. Vector Space Model dengan Label

**Vector Space Model (VSM)** = tabel yang barisnya dokumen dan kolomnya fitur. Di sini ada dua versi, keduanya dilengkapi kolom `label_kelas` dan `split_data` (train / test):

- **One-Hot**: kolom = kata (nilai 0/1).
- **Skip-gram**: kolom = dimensi embedding (rata-rata vektor kata).

````python
tokens_p1 = prep_results[MODEL_PREP].tolist()
split_data = np.where(np.isin(np.arange(len(df_berita)), idx_train), "train", "test")

# (a) VSM One-Hot  (kosakata dari data latih saja -> pakai `onehot` dari bagian 7.3)
X_oh = onehot.transform(tokens_p1)
terms = onehot.get_feature_names_out()
df_vsm_onehot = pd.DataFrame(X_oh.toarray(), columns=terms)
df_vsm_onehot.insert(0, "label_kelas", df_berita["label"].values)
df_vsm_onehot.insert(1, "split_data", split_data)
print("Ukuran VSM One-Hot :", df_vsm_onehot.shape)

top_terms = terms[np.argsort(np.asarray(X_oh.sum(axis=0)).ravel())[::-1][:12]]   # 12 kata paling sering muncul
display(df_vsm_onehot[["label_kelas", "split_data", *top_terms]].head(8))

# (b) VSM Skip-gram  (vektor dokumen dari model final)
X_sg = np.array([dokumen_ke_vektor(t, w2v_final) for t in tokens_p1])
df_vsm_skipgram = pd.DataFrame(X_sg, columns=[f"dim_{i+1}" for i in range(X_sg.shape[1])])
df_vsm_skipgram.insert(0, "label_kelas", df_berita["label"].values)
df_vsm_skipgram.insert(1, "split_data", split_data)
print("\nUkuran VSM Skip-gram :", df_vsm_skipgram.shape)
display(df_vsm_skipgram.iloc[:8, :8].round(4))

# (c) Proyeksi 2D (PCA): apakah dua kelas terpisah di ruang vektor?
Z = PCA(n_components=2, random_state=SEED).fit_transform(X_sg)
plt.figure(figsize=(6.5, 5))
sns.scatterplot(x=Z[:, 0], y=Z[:, 1], hue=df_berita["label"].values,
                style=split_data, palette="Set2", s=60)
plt.title("Ruang Vektor Dokumen (Skip-gram) — PCA 2D")
plt.xlabel("PC 1"); plt.ylabel("PC 2")
plt.show()
````
